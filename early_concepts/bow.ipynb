# Corpus


We will take a toy text file which contains labeled text messages [spam or ham] separated by a tab. We first split to get the labels and the text, then clena up the text to remove extra spaces, punctuation, etc. The text is then split into words, and then a set of unique words which forms the corpus. For ease, we take the first three hundred or so of the texts.  

```python
import urllib.request

url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/sms.tsv"
filename = "sms_spam.tsv"
urllib.request.urlretrieve(url, filename)

# Read lines into a corpus list
with open(filename, "r", encoding="utf-8") as f:
    # Each line has a label and text separated by a tab
    lines = f.readlines()

# len(lines)
```

```python
import pandas as pd

labels = [line.split("\t")[0].strip() for line in lines]
text = [line.split("\t")[1].strip() for line in lines]

# check
# text[9]  ## spam example
# labels[9]

# text[4]  ## ham (not spam) example
# labels[4]
```

```python
# Converting labels and text as cols in DF

data = pd.DataFrame({'Label': labels, 'Text':text})

# data.shape

# data.Label.value_counts() # approx 7:1

# data.sample(7)
```

```python
# taking the first few sentences to define our corpus
fulltext = " ".join(text[:350])
# fulltext[150:275]

# strip punctuation
import string
clean = fulltext.translate(str.maketrans("", "", string.punctuation))

# print(clean[100:175])

words = clean.lower().split()
vocab = set(clean.lower().split())
```

```python
# fulltext
# clean
```

```python
## str len of all sentences combined; total words in all sentences; unique number of words

print(len(fulltext), len(clean), len(words), len(vocab))
```

```python
ordered = sorted(vocab)

word_dict = {v:k for k,v in enumerate(ordered)}


print(len(ordered), len(word_dict))
```

We now have a word-index mapping that can be used as a reference to construct the same for any sentence or document from hereon.

# <b> BoW


Mapping unique words to fixed vector positions leads directly to two core ideas:

1. One-Hot Vectors: Representing a single word as a vector of size
$N$ (where $N$ is vocabulary size) with a single 1 and the rest 0s.

2. Bag of Words (BoW): Representing an entire document by combining those word representations into a single fixed-length vector.  


Its biggest flaw: word order is destroyed. "Dog bites man" and "Man bites dog" result in the exact same vector representation.
